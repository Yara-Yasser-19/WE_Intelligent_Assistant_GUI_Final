# Telecom Egypt Intelligent Assistant
Desktop GUI + Local RAG Proof of Concept.


## Architecture
Knowledge → cleaning → chunking → retrieval → grounded answer → source citation.


```python
from pathlib import Path
from rag_engine import RAGEngine
```

```python
engine = RAGEngine(Path.cwd())
print('Chunks:', engine.chunk_count())
print('Sources:', engine.source_summary())
```

## Retrieval Example


```python
result = engine.answer('What is Nitro Prime 150?')
print(result['answer'])
print('Sources:', result['sources'])
```

## Production Upgrade
For production, replace the lightweight keyword retriever with multilingual embeddings and a vector store such as FAISS or Chroma. An on-premises LLM can then generate natural answers from retrieved context while keeping source citations.
